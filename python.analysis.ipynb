```python
import pandas as pd
from datetime import timedelta
df = pd.read_csv(r'C:\Users\A\Downloads\enterprise-customers-1000.csv')
```

```python
print("Number of unique industries:", df['Industry'].nunique())
print("\nTop 5 industries:")
print(df['Industry'].value_counts().head(5))
```

```python
df.columns = df.columns.str.strip().str.lower()
print(df.columns.tolist())

```

```python
print("Total Employees:", df['number of employees'].sum())
print("Avg Revenue:", df['annual revenue'].mean())
```

```python
df[(df['number of employees'] > 8000) & (df['annual revenue'] < 500)]
```

```python
print("Missing Phone Secondary:",)
df['phone secondary'].isna().sum()
```

```python
print("Negative:",)
print (df[df['annual revenue'] < 0])
```

```python
print("By Country:",)
print(df.groupby('country')['contract value'].sum().sort_values(ascending=False).head())
```

```python
print("Support Level:")
print(df.groupby('support level')['contract value'].mean())

```

```python
print("By Department:")
print(df.groupby('department')['contract value'].mean().sort_values(ascending=False))
```

```python
df['contract start date'] = pd.to_datetime(df['contract start date'],errors='coerce')
df['contract end date'] = pd.to_datetime(df['contract end date'],errors='coerce')
df['last contact date'] = pd.to_datetime(df['last contact date'],errors='coerce')
df[ 'next follow up date'] = pd.to_datetime(df[ 'next follow up date'],errors='coerce')

```

```python
today = pd.to_datetime('today').normalize()
print(today)
```

```python
print(df[['contract start date', 'contract end date']].dtypes)
df['Duration Days'] = (df['contract end date'] - df['contract start date']).dt.days
df[df['contract start date'].isna() | df['contract end date'].isna()]


print('Longest Contract:')
print( df.sort_values('Duration Days', ascending=False)
      [['company name', 'Duration Days']].head(1))
```

```python
df['contract end date'] = pd.to_datetime(df['contract end date'])
```

```python

next_90 = today + timedelta(days=90)
renewal = df[(df['contract end date'] >= today) &
     (df['contract end date'] <= next_90)]
print(f"Renewal needed: {len(renewal)}")

```

```python
print(df['next follow up date'].dtype)
print(df['last contact date'].dtype)
print(type(today))
print(df['next follow up date'].head())
```

```python

overdue = df[(df['next follow up date'] < today) &
    ((today - df['last contact date']).dt.days > 30)]
print(f"overdue follow ups: {len(overdue)}")
```

```python
df['Revenue per Employee'] =df['annual revenue'] / df['number of employees']
Top_10 =  df.sort_values('Revenue per Employee', ascending=False).head(10)
print(Top_10[['company name','Revenue per Employee']])
```

```python
today = pd.Timestamp.today()
```

```python

df['last contact date'] = pd.to_datetime(df['last contact date'])
df['Churn Risk'] = (((today - df['last contact date']).dt.days > 180) &
    (df['support level'] == 'standard'))
at_risk = df[df['Churn Risk']]
print(f'Churn Risk Customer: {len(at_risk)}')
```

```python
industry = df.groupby('industry').agg(
    total_companies=('company name', 'count'),
    total_revenue=('annual revenue', 'sum')
).round(2)

print(industry)
```

```python
country = df.groupby('country').agg(
    total_companies=('company name', 'count'),
    total_revenue=('annual revenue', 'sum')
).round(2)

print(country)
```

```python
manager = df.groupby('account manager').agg(
    total_client=('company name', 'count'),
    total_revenue=('annual revenue', 'sum'),
    churn_count=('churn risk', 'sum')
).round(2)

print(manager)
```

```python
import matplotlib.pyplot as plt
top_10 = (df.groupby('industry')['contract value'].sum()
      .sort_values(ascending=False)
      .head(10)
      .sort_values())
plt.figure(figsize=(10, 5))
top_10.plot(kind='barh')
plt.title('Top 10 Contract Value by Industry')
plt.xlabel('Contract Value')
plt.ylabel('Industry')
plt.tight_layout()
plt.show()
```
